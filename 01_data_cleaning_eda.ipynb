```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("D:\Bank Customer Churn\Bank_Churn.csv")
```

```python
print(df.shape)
print(df.info())
print(df.isnull().sum())
print(df.describe())
```

```python
missing = df.isnull().sum()
print(missing)
```

```python
print("DATA TYPES:")
print("="*50)
print(df.dtypes)
print("\n" + "="*50)
print("\nDetailed Info:")
df.info()
```

```python
# KPI Metrics
total_customers = len(df)
churned_customers = df['Exited'].sum()
churn_rate = (churned_customers / total_customers) * 100
retention_rate = 100 - churn_rate

print(f"Total Customers: {total_customers}")
print(f"Churned: {churned_customers}")
print(f"Churn Rate: {churn_rate:.2f}%")
print(f"Retention Rate: {retention_rate:.2f}%")

# Create a summary metrics dataframe (for Power BI KPI cards)
metrics = pd.DataFrame({
    'Metric': ['Total Customers', 'Total Churned', 'Churn Rate %', 'Retention Rate %'],
    'Value': [total_customers, churned_customers, round(churn_rate, 2), round(retention_rate, 2)]
})
```

```python
# Age Group
df['AgeGroup'] = pd.cut(df['Age'], 
                        bins=[0, 30, 45, 60, 100], 
                        labels=['18-30', '31-45', '46-60', '61+'])

# Credit Score Band
df['CreditScoreBand'] = pd.cut(df['CreditScore'],
                               bins=[0, 500, 700, 900],
                               labels=['Poor (0-500)', 'Good (500-700)', 'Excellent (700+)'])

# Tenure Group
df['TenureGroup'] = pd.cut(df['Tenure'],
                           bins=[0, 2, 5, 10, 15],
                           labels=['0-2 Years', '2-5 Years', '5-10 Years', '10+ Years'])

# Balance Category
df['BalanceCategory'] = pd.cut(df['Balance'],
                               bins=[0, 50000, 100000, 250000],
                               labels=['$0-50K', '$50K-100K', '$100K+'])

print("Segmentation columns created!")
print(df[['AgeGroup', 'CreditScoreBand', 'TenureGroup', 'BalanceCategory']].head())
```

```python
# Churn by Geography
churn_by_geo = df.groupby('Geography').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_geo.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_geo['ChurnRate'] = churn_by_geo['ChurnRate'] 
```

```python
# Churn by Gender
churn_by_gender = df.groupby('Gender').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_gender.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_gender['ChurnRate'] = churn_by_gender['ChurnRate'] 
```

```python
# Churn by Age Group
churn_by_age = df.groupby('AgeGroup').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_age.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_age['ChurnRate'] = churn_by_age['ChurnRate'] 
```

```python
churn_by_credit = df.groupby('CreditScoreBand').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_credit.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_credit['ChurnRate'] = churn_by_credit['ChurnRate'] 
```

```python
# Churn by Membership Status (IsActiveMember)
churn_by_member = df.groupby('IsActiveMember').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_member.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_member['ChurnRate'] = churn_by_member['ChurnRate'] 
churn_by_member.index = ['Active', 'Not Active']
```

```python
# Churn by Balance (you'll use this for the Balance chart)
churn_by_balance = df.groupby('BalanceCategory').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_balance.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_balance['ChurnRate'] = churn_by_balance['ChurnRate'] 
```

```python
# Churn by Tenure
churn_by_tenure = df.groupby('Tenure').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_tenure.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_tenure['ChurnRate'] = churn_by_tenure['ChurnRate'] 
```

```python
 # Churn by Number of Products
churn_by_products = df.groupby('NumOfProducts').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_products.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_products['ChurnRate'] = churn_by_products['ChurnRate'] 
```

```python
# Churn by Estimated Salary (create bins)
df['SalaryBand'] = pd.cut(df['EstimatedSalary'],
                          bins=[0, 50000, 100000, 150000, 200000],
                          labels=['$0-50K', '$50K-100K', '$100K-150K', '$150K+'])
churn_by_salary = df.groupby('SalaryBand').agg({
    'Exited': ['sum', 'count', 'mean']
}).round(4)
churn_by_salary.columns = ['Churned', 'Total', 'ChurnRate']
churn_by_salary['ChurnRate'] = churn_by_salary['ChurnRate'] 
```

```python
with pd.ExcelWriter('bank_churn_processed.xlsx', engine='openpyxl') as writer:
    # Main dataset (for detailed analysis)
    df.to_excel(writer, sheet_name='Raw Data', index=False)
    
    # KPI metrics
    metrics.to_excel(writer, sheet_name='KPI Summary', index=False)
    
    # Segmentation analysis
    churn_by_geo.reset_index().to_excel(writer, sheet_name='By Geography', index=False)
    churn_by_gender.reset_index().to_excel(writer, sheet_name='By Gender', index=False)
    churn_by_age.reset_index().to_excel(writer, sheet_name='By Age', index=False)
    churn_by_credit.reset_index().to_excel(writer, sheet_name='By Credit Score', index=False)
    churn_by_member.reset_index().to_excel(writer, sheet_name='By Member Status', index=False)
    churn_by_balance.reset_index().to_excel(writer, sheet_name='By Balance', index=False)
    churn_by_tenure.reset_index().to_excel(writer, sheet_name='By Tenure', index=False)
    churn_by_products.reset_index().to_excel(writer, sheet_name='By Products', index=False)
    churn_by_salary.reset_index().to_excel(writer, sheet_name='By Salary', index=False)
```
